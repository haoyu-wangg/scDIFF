```python
import matplotlib.pyplot as plt
import numpy as np

# Set font to Times New Roman
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 14

# Data
dataset_pairs = [
    'BoneMarrowA_BoneMarrowB',
    'BoneMarrowB_BoneMarrowA', 
    'LargeIntestineA_LargeIntestineB',
    'LargeIntestineB_LargeIntestineA',
    'LungA_LungB',
    'LungB_LungA',
    'MosA1_MosA2',
    'MosA2_MosA1',
    'MosM1_MosM2',
    'MosM2_MosM1',
    'MosP1_MosP2',
    'MosP2_MosP1',
    'WholeBrainA_WholeBrainB',
    'WholeBrainB_WholeBrainA'
]

sango_times = [206.07, 211.27, 154.48, 156.43, 219.13, 221.34, 
               317.38, 314.47, 313.57, 317.79, 318.29, 319.45, 
               211.74, 206.30]

scdiff_times = [66.37, 52.79, 32.98, 38.13, 64.65, 63.92,
                75.05, 60.10, 51.65, 54.55, 83.74, 53.06,
                57.19, 48.41]

# Create figure and axis
fig, ax = plt.subplots(figsize=(12, 8))

# Set y-axis positions
y_pos = np.arange(len(dataset_pairs))
bar_height = 0.35

# Draw horizontal bar chart
bars1 = ax.barh(y_pos - bar_height/2, sango_times, bar_height, 
                label='SANGO', color='#ff7f0e', alpha=0.8)
bars2 = ax.barh(y_pos + bar_height/2, scdiff_times, bar_height,
                label='scDIFF', color='#2ca02c', alpha=0.8)

# Set y-axis labels
ax.set_yticks(y_pos)
ax.set_yticklabels(dataset_pairs)

# Set axis labels and title
ax.set_xlabel('Execution Time (seconds)', fontsize=18, fontweight='bold')
# ax.set_ylabel('Dataset Pairs', fontsize=18, fontweight='bold')
# ax.set_title('Execution Time Comparison', 
#              fontsize=18, fontweight='bold', pad=20)

# Add legend
ax.legend(loc='lower right', fontsize=16)

# Add grid lines
ax.grid(axis='x', alpha=0.3, linestyle='--')

# Add value labels on the bars
for i, (bar1, bar2) in enumerate(zip(bars1, bars2)):
    # SANGO values
    width1 = bar1.get_width()
    ax.text(width1 + 5, bar1.get_y() + bar1.get_height()/2, 
            f'{width1:.2f}', ha='left', va='center', fontsize=14)
    
    # scDIFF values
    width2 = bar2.get_width()
    ax.text(width2 + 5, bar2.get_y() + bar2.get_height()/2,
            f'{width2:.2f}', ha='left', va='center', fontsize=14)

# Adjust layout
plt.tight_layout()

# Invert y-axis order (so the first dataset is on top)
ax.invert_yaxis()

# Set x-axis range
ax.set_xlim(0, max(max(sango_times), max(scdiff_times)) * 1.15)

# Save figure
plt.savefig('execution_time_comparison.svg', 
           format='svg', 
           bbox_inches='tight',
           facecolor='white',      # Background color
           edgecolor='none',       # Border color
           transparent=False)      # Whether background is transparent

# Show figure
plt.show()
```
